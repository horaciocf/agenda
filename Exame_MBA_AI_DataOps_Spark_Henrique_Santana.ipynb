```python
import time
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, count
```

```python
# Inicializa uma SparkSession
spark = (
    SparkSession.builder
                .appName("Desafio Processamento de Dados")
                .master("local[*]")
                .getOrCreate()
)
```

```python
# Carrega o dataset
file_url = "https://gist.githubusercontent.com/designernatan/27da044c6dc823f7ac7fe3a01f4513ed/raw/d15b5c7d7a5efb38750b16ec935fc126ec9a6e79/vgsales.csv"
pd_df = pd.read_csv(file_url)
vgsales_df = spark.createDataFrame(pd_df)
```

```python
# Exibe algumas linhas do DataFrame
vgsales_df.show(5)
```

```python
# Insights dos jogos da Sony
sony_releases = (
    vgsales_df.select("Year", "Platform", "Name")
              .filter(vgsales_df.Platform.isin(["PS", "PS2", "PS3", "PS4"]))
              .groupBy("Year", "Platform")
              .agg(count("Name").alias("Releases"))
              .orderBy("Year", ascending=True)
)
```

```python
sony_releases.show()
```

```python
# Insights dos jogos da Microsoft
microsoft_releases = (
    vgsales_df.select("Year", "Platform", "Name")
              .filter(vgsales_df.Platform.isin(["XB", "X360", "XOne"]))
              .groupBy("Year", "Platform")
              .agg(count("Name").alias("Releases"))
              .orderBy("Year", ascending=True)
)
```

```python
microsoft_releases.show()
```

```python
# Insights dos jogos da Nintendo
nintendo_releases = (
    vgsales_df.select("Year", "Platform", "Name")
              .filter(vgsales_df.Platform.isin(["NES", "SNES", "N64", "GC", "Wii", "WiiU"]))
              .groupBy("Year", "Platform")
              .agg(count("Name").alias("Releases"))
              .orderBy("Year", ascending=True)
)
```

```python
nintendo_releases.show()
```
